```python
import tensorflow as tf
```

```python
# Does tensorflow see our GPU?
tf.config.list_physical_devices('GPU')
```

## Load MobileNetV3Small

```python
mobilenet = tf.keras.applications.MobileNetV3Small(input_shape=(224,224,3), weights='imagenet')
```

## Convert to TFLite

```python
converter = tf.lite.TFLiteConverter.from_keras_model(mobilenet)
#Enable optimization
converter.optimizations = [tf.lite.Optimize.DEFAULT]
tflite_model = converter.convert()
```

## Save TFLite Model

```python
with open('mobilenetv3_small.tflite', 'wb') as f:
    f.write(tflite_model)
```
